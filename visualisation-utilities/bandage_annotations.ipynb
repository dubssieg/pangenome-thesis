```python
gfa_path:str = "/home/sidubois/Data_pangenomes/graph_comparison_paper/mg_yeast_1.gfa"
reference:str = "CASBJU01"
alternative:str = "CASBIT01"
```

```python
reference_set:set = set()
alternative_set:set = set()
sharing:dict = dict()
sharing_paths:dict = dict()

with open(gfa_path,'r',encoding='utf-8') as gfa_reader:
    for line in gfa_reader:
        if line.startswith('P'):
            nodes = [x[:-1] for x in line.strip().split('\t')[2].split(',')]
            for node in nodes:
                if node in sharing:
                    sharing[node] += 1
                else:
                    sharing[node] = 1
            nodes = set(nodes)
            for node in nodes:
                if node in sharing_paths:
                    sharing_paths[node] += 1
                else:
                    sharing_paths[node] = 1
            if reference in line.strip().split('\t')[1]:
                reference_set = set([x[:-1] for x in line.strip().split('\t')[2].split(',')])
            elif alternative in line.strip().split('\t')[1]:
                alternative_set = set([x[:-1] for x in line.strip().split('\t')[2].split(',')])

intersection_set:set = reference_set.intersection(alternative_set)


with open("MC_annotations.csv", 'w+',encoding='utf-8') as annot:
    annot.write("Name,Label,Status,Sharing,Paths" + '\n')
    with open(gfa_path,'r',encoding='utf-8') as gfa_reader:
        for line in gfa_reader:
            if line.startswith('S'):
                status:str = "∅"
                if line.strip().split('\t')[1] in intersection_set:
                    status = 'N'
                elif line.strip().split('\t')[1] in reference_set:
                    status = 'R'
                elif line.strip().split('\t')[1] in alternative_set:
                    status = 'A'
                annot.write(line.strip().split('\t')[1] + "," + line.strip().split('\t')[2] + ',' + status +',' + str(sharing[line.strip().split('\t')[1]]) +',' + str(sharing_paths[line.strip().split('\t')[1]]) +'\n')
```
